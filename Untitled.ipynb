```python
import torch
from torch.utils.data import Dataset, DataLoader
import librosa
import numpy as np
import glob
import os

class MIMIIDataset(Dataset):
    def __init__(self, file_list, sr=16000, n_mels=128):
        self.file_list = file_list
        self.sr = sr
        self.n_mels = n_mels
        
    def __len__(self):
        return len(self.file_list)
        
    def __getitem__(self, idx):
        file_path = self.file_list[idx]
        
        # Load audio
        y, _ = librosa.load(file_path, sr=self.sr)
        
        # Extract Mel-Spectrogram and convert to Log scale
        mel_spec = librosa.feature.melspectrogram(y=y, sr=self.sr, n_mels=self.n_mels, n_fft=2048, hop_length=512)
        log_mel = librosa.power_to_db(mel_spec, ref=np.max)
        
        # Min-Max normalization to [0, 1]
        log_mel = (log_mel - log_mel.min()) / (log_mel.max() - log_mel.min() + 1e-8)
        
        # PyTorch expects shape: [Channels, Height, Width] -> [1, 128, 313]
        tensor_spec = torch.tensor(log_mel, dtype=torch.float32).unsqueeze(0)
        return tensor_spec

# Get list of normal training files
data_path = "./data/fan"
train_files = glob.glob(os.path.join(data_path, "**", "train", "*.wav"), recursive=True)

# Create Dataset and DataLoader
train_dataset = MIMIIDataset(train_files)
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True, drop_last=True)

print(f"✅ DataLoader created successfully!")
print(f"  └── Total training batches available: {len(train_loader)}")
```

```python
import torch.nn as nn

class ConvAutoencoder(nn.Module):
    def __init__(self):
        super(ConvAutoencoder, self).__init__()
        
        # --- ENCODER ---
        self.encoder = nn.Sequential(
            nn.Conv2d(1, 16, kernel_size=3, stride=2, padding=1),  # Output: [16, 64, 157]
            nn.ReLU(True),
            nn.Conv2d(16, 32, kernel_size=3, stride=2, padding=1), # Output: [32, 32, 79]
            nn.ReLU(True),
            nn.Conv2d(32, 64, kernel_size=3, stride=2, padding=1), # Output: [64, 16, 40] (Bottleneck)
            nn.ReLU(True)
        )
        
        # --- DECODER ---
        self.decoder = nn.Sequential(
            nn.ConvTranspose2d(64, 32, kernel_size=3, stride=2, padding=1, output_padding=(1, 0)), # Output: [32, 32, 79]
            nn.ReLU(True),
            nn.ConvTranspose2d(32, 16, kernel_size=3, stride=2, padding=1, output_padding=(1, 1)), # Output: [16, 64, 158] -> adjusted
            nn.ReLU(True),
            nn.ConvTranspose2d(16, 1, kernel_size=3, stride=2, padding=1, output_padding=(1, 0)),  # Output: [1, 128, 314]
            nn.Sigmoid() # Keeps outputs bounded between 0 and 1
        )

    def forward(self, x):
        encoded = self.encoder(x)
        decoded = self.decoder(encoded)
        # Crop or adjust slightly if needed to match exact input width (313)
        return decoded[:, :, :, :313]

# Test the model with a single batch
model = ConvAutoencoder()
sample_batch = next(iter(train_loader))
output = model(sample_batch)

print("🧠 Autoencoder Architecture Initialized!")
print(f"  └── Input Batch Shape: {sample_batch.shape}")
print(f"  └── Output Reconstructed Shape: {output.shape}")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

# 1. Set up device (uses GPU if available, otherwise CPU)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"🚀 Using device: {device}")

# Move model to device
model = ConvAutoencoder().to(device)

# 2. Define Loss function and Optimizer
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# 3. Training Hyperparameters
epochs = 5  # 5 to 10 epochs is usually plenty for a quick baseline on normal fan audio

print("Starting training...")
for epoch in range(epochs):
    model.train()
    running_loss = 0.0
    
    for batch in train_loader:
        # Move batch to device
        batch = batch.to(device)
        
        # Zero gradients
        optimizer.zero_grad()
        
        # Forward pass (reconstruct the spectrogram)
        outputs = model(batch)
        
        # Calculate loss (Error between input and output)
        loss = criterion(outputs, batch)
        
        # Backward pass and optimize
        loss.backward()
        optimizer.step()
        
        running_loss += loss.item() * batch.size(0)
        
    epoch_loss = running_loss / len(train_loader.dataset)
    print(f"Epoch [{epoch+1}/{epochs}] | Loss (MSE): {epoch_loss:.6f}")

print("\n🎉 Training Complete! Your model now knows what a normal fan sounds like.")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import librosa
import numpy as np
import glob
import os

# 1. Dataset with Fixed Global Scaling
class MIMIIDatasetFixed(Dataset):
    def __init__(self, file_list, sr=16000, n_mels=128):
        self.file_list = file_list
        self.sr = sr
        self.n_mels = n_mels
        
    def __len__(self):
        return len(self.file_list)
        
    def __getitem__(self, idx):
        y, _ = librosa.load(self.file_list[idx], sr=self.sr)
        mel_spec = librosa.feature.melspectrogram(y=y, sr=self.sr, n_mels=self.n_mels, n_fft=2048, hop_length=512)
        log_mel = librosa.power_to_db(mel_spec, ref=np.max)
        
        # Fixed global scaling: Map [-80dB, 0dB] smoothly to [0, 1]
        log_mel = np.clip(log_mel, -80.0, 0.0) / 80.0 + 1.0 
        return torch.tensor(log_mel, dtype=torch.float32).unsqueeze(0)

# Load training data
data_path = "./data/fan"
train_files = glob.glob(os.path.join(data_path, "**", "train", "*.wav"), recursive=True)

train_dataset = MIMIIDatasetFixed(train_files)
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True, drop_last=True)

# 2. Setup Device & Model
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = ConvAutoencoder().to(device)
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# 3. Train for 10 Epochs
epochs = 10
print(f"🚀 Training with fixed global scaling on {device}...")
for epoch in range(epochs):
    model.train()
    running_loss = 0.0
    for batch in train_loader:
        batch = batch.to(device)
        optimizer.zero_grad()
        outputs = model(batch)
        loss = criterion(outputs, batch)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * batch.size(0)
        
    epoch_loss = running_loss / len(train_loader.dataset)
    print(f"Epoch [{epoch+1}/{epochs}] | Loss: {epoch_loss:.6f}")

print("\n🎉 Retraining complete!")

# Save this newly trained fixed model
os.makedirs("./models", exist_ok=True)
torch.save(model.state_dict(), "./models/fan_autoencoder_fixed.pth")
print("✅ Saved fixed model weights to ./models/fan_autoencoder_fixed.pth")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import librosa
import numpy as np
import glob
import os
import matplotlib.pyplot as plt
from sklearn.metrics import roc_auc_score, roc_curve

# --- 1. RETRAIN & TRACK LOSS CURVE ---
class MIMIIDatasetFixed(Dataset):
    def __init__(self, file_list, sr=16000, n_mels=128):
        self.file_list = file_list
        self.sr = sr
        self.n_mels = n_mels
        
    def __len__(self):
        return len(self.file_list)
        
    def __getitem__(self, idx):
        y, _ = librosa.load(self.file_list[idx], sr=self.sr)
        mel_spec = librosa.feature.melspectrogram(y=y, sr=self.sr, n_mels=self.n_mels, n_fft=2048, hop_length=512)
        log_mel = librosa.power_to_db(mel_spec, ref=np.max)
        log_mel = np.clip(log_mel, -80.0, 0.0) / 80.0 + 1.0 
        return torch.tensor(log_mel, dtype=torch.float32).unsqueeze(0)

data_path = "./data/fan"
train_files = glob.glob(os.path.join(data_path, "**", "train", "*.wav"), recursive=True)
train_dataset = MIMIIDatasetFixed(train_files)
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True, drop_last=True)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = ConvAutoencoder().to(device)
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

epochs = 10
epoch_losses = []

print(f"🚀 Training model and tracking loss...")
for epoch in range(epochs):
    model.train()
    running_loss = 0.0
    for batch in train_loader:
        batch = batch.to(device)
        optimizer.zero_grad()
        outputs = model(batch)
        loss = criterion(outputs, batch)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * batch.size(0)
        
    epoch_loss = running_loss / len(train_loader.dataset)
    epoch_losses.append(epoch_loss)
    print(f"Epoch [{epoch+1}/{epochs}] | Loss: {epoch_loss:.6f}")


# --- 2. EVALUATE ALL TEST FILES FOR FULL AUC ---
model.eval()
criterion_eval = nn.MSELoss(reduction='mean')

test_normal_files = glob.glob(os.path.join(data_path, "**", "test", "*normal*.wav"), recursive=True)
test_anomaly_files = glob.glob(os.path.join(data_path, "**", "test", "*anomaly*.wav"), recursive=True)

print(f"\nEvaluating across {len(test_normal_files)} normal and {len(test_anomaly_files)} anomaly test files...")

def get_all_errors(file_list):
    errors = []
    with torch.no_grad():
        for f in file_list:
            y, sr = librosa.load(f, sr=16000)
            mel = librosa.feature.melspectrogram(y=y, sr=sr, n_mels=128, n_fft=2048, hop_length=512)
            log_mel = librosa.power_to_db(mel, ref=np.max)
            log_mel = np.clip(log_mel, -80.0, 0.0) / 80.0 + 1.0
            
            tensor = torch.tensor(log_mel, dtype=torch.float32).unsqueeze(0).unsqueeze(0).to(device)
            recon = model(tensor)
            loss = criterion_eval(recon, tensor).item()
            errors.append(loss)
    return np.array(errors)

normal_errors = get_all_errors(test_normal_files)
anomaly_errors = get_all_errors(test_anomaly_files)

# Combine for AUC calculation (0 = Normal, 1 = Anomaly)
y_true = [0] * len(normal_errors) + [1] * len(anomaly_errors)
y_scores = list(normal_errors) + list(anomaly_errors)

auc_score = roc_auc_score(y_true, y_scores)
print(f"\n🏆 Official Test AUC-ROC Score: {auc_score:.4f}")


# --- 3. GENERATE VISUAL GRAPHS ---
plt.figure(figsize=(14, 5))

# Plot 1: Training Loss Curve
plt.subplot(1, 2, 1)
plt.plot(range(1, epochs + 1), epoch_losses, marker='o', color='b', linewidth=2)
plt.title('Training Loss Curve (Autoencoder Convergence)', fontsize=12, fontweight='bold')
plt.xlabel('Epoch', fontsize=10)
plt.ylabel('MSE Loss', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.6)

# Plot 2: Reconstruction Error Distribution
plt.subplot(1, 2, 2)
plt.hist(normal_errors, bins=30, alpha=0.6, label='Normal Fans', color='green', edgecolor='black')
plt.hist(anomaly_errors, bins=30, alpha=0.6, label='Anomalous Fans', color='red', edgecolor='black')
plt.title('Reconstruction Error Distribution', fontsize=12, fontweight='bold')
plt.xlabel('Mean Squared Error (MSE)', fontsize=10)
plt.ylabel('Sample Count', fontsize=10)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
import onnxruntime as ort
import numpy as np

# 1. Automatically detect available hardware (falls back to CPU if no GPU)
available_providers = ort.get_available_providers()
print(f"🔌 Available ONNX Providers: {available_providers}")

# 2. Load the ONNX model device-agnostically
onnx_path = "./models/fan_autoencoder.onnx"
session = ort.InferenceSession(onnx_path, providers=available_providers)

# 3. Prepare a sample input (ONNX expects NumPy arrays instead of PyTorch Tensors)
dummy_input = np.random.randn(1, 1, 128, 313).astype(np.float32)

# 4. Run inference
input_name = session.get_inputs()[0].name
output_name = session.get_outputs()[0].name

outputs = session.run([output_name], {input_name: dummy_input})
reconstructed_output = outputs[0]

print("✅ Device-agnostic ONNX inference successful!")
print(f"  └── Reconstructed Output Shape: {reconstructed_output.shape}")
```
